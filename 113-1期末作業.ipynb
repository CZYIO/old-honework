```python
import random
#定義玩家類別
class 玩家:
    def __init__(self,名稱):
        self.名稱=名稱
        self.生命值=100
        self.攻擊力=10
        self.防禦力 = 5
        self.等級=1
        self.經驗值=0
        self.技能點=0
        self.背包=[]
        self.暴擊率 = 0.5
        self.暴擊加成= 1.5
        #技能系統
        self.技能= {"火球術":15,"治癒術":20}

    def 攻擊敵人(self,敵人):
        暴擊=random.random()<self.暴擊率
        if 暴擊:
            傷害=max(0,int(self.攻擊力*self.暴擊加成-敵人.防禦力))
            print(f"💥暴擊!{self.名稱}對{敵人.名稱}造成了{傷害}點傷害!")
        else:
            傷害=max(0,(self.攻擊力-敵人.防禦力))
            print(f"⚔{self.名稱}對{敵人.名稱}造成了{傷害}點傷害!")
        敵人.生命值-=傷害

    def 使用技能(self,敵人):
        print("📜可用技能：")
        for 編號,(技能名稱,效果) in enumerate(self.技能.items(),start=1):
            print(f"{編號}.{技能名稱}-{效果}點效果")

            選擇 = int(input("請選擇要使用的技能編號:"))-1
            技能名稱 =list(self.技能.keys())[選擇]
            技能效果 = self.技能[技能名稱]
            if 技能名稱=="火球術":
                敵人.生命值-=技能效果
                print(f"🔥{self.名稱}使用{技能名稱},對{敵人.名稱}造成了{技能效果}點傷害!")
            elif 技能名稱 =="治癒術":
                self.生命值 +=技能效果
                print(f"✨{self.名稱}使用{技能名稱},恢復了{技能效果}點生命值!")

    def 是否存活(self):
        return self.生命值>0

    def 獲得經驗值(self,獎勵經驗):
        self.經驗值+=獎勵經驗
        print(f"✨{self.名稱}獲得了{獎勵經驗}點經驗值!")
        if self.經驗值>=self.等級*20:
            self.升級()

    def 升級(self):
        self.等級+=1
        self.生命值+=20
        self.攻擊力+=5
        self.防禦力+=3
        self.技能點+=1
        self.暴擊率+=0.05
        print(f"🎉恭喜!{self.名稱}升到了{self.等級}!技能點數增加為{self.技能點}點!")

    def 使用道具(self):
        if not self.背包:
            print("👜背包是空的,無法使用道具!")
            return
        print("📜道具列表：")
        for 編號,道具 in enumerate(self.背包,start=1):
            print(f"{編號}.{道具['名稱']}-{道具['效果']}")
        選擇 =int(input("請選擇要使用的道具編號:"))-1
        if 0 <=選擇<len(self.背包):
            道具=self.背包.pop(選擇)
            if 道具["類型"]=="恢復":
                self.生命值+=道具["數值"]
                print(f"💊{道具['名稱']}使用成功,恢復了{道具['數值']}點生命值!")
            else:
                print("❌無效的選擇,請重新操作!")
#定義敵人類別
class 敵人:
   def __init__(self,名稱,攻擊力,防禦力,獎勵經驗):
    self.名稱=名稱
    self.生命值=100
    self.攻擊力=攻擊力
    self.防禦力=防禦力
    self.獎勵經驗=獎勵經驗

   def 攻擊玩家(self,玩家):
    傷害=max(0,self.攻擊力-玩家.防禦力)
    玩家.生命值-=傷害
    print(f"🔥{self.名稱}對{玩家.名稱}造成了{傷害}點傷害!")

   def 是否存活(self):
    return self.生命值>0

   def 隨機事件(玩家):
    事件機率=random.random()
    if 事件機率<0.3:
        print("🎲 你在路上發現了一瓶神秘的恢復藥水!")
        玩家.生命值+=20
        print(f"✨生命值增加了20點,目前生命值:{玩家.生命值}")
    elif 事件機率< 0.6:
        print("🎲 遭遇了野狼襲擊!你失去了15點生命值。")
        玩家.生命值-=15
    else:
        print("🎲 幸運的日子!你找到了一袋黃金 沒有任何負面影響。")

#遊戲主程式
   def 主程式():
    print("⭐歡迎來到勇者闖關遊戲!")
    玩家名稱=input("🛡請輸入你的角色名稱:")
    玩家角色=玩家(玩家名稱)

    敵人列表 = [
    敵人("史萊姆🟢",30,5,2,10),
    敵人("哥布林🧞‍♀️",50,8,4,20),
    敵人("巨型惡龍🐉",100,15,10,50)
    ]
    for 關卡,敵人角色 in enumerate(敵人列表,start=1):
        print(f"\n第{關卡}關:{敵人角色.名稱}出現了!")
        隨機事件(玩家角色)

    while 玩家角色.是否存活()and 敵人角色.是否存活():
         print(f"\n❤{玩家角色.名稱}的生命值:{玩家角色.生命值}")
         print(f"\n💔{敵人角色.名稱}的生命值:{敵人角色.生命值}")

    行動=input("請選擇行動(1:攻擊⚔,2:防禦🛡,3:補血💊,4:使用技能📜,5:使用道具👜)")
    if 行動=="1":
        玩家角色.攻擊敵人(敵人角色)
    elif 行動=="2":
        print(f"{玩家角色.名稱}採取了防禦行動!")
        玩家角色.防禦力+=2
    elif 行動=="3":
        補血量=random.randint(10,20)
        玩家角色.生命值+=2
        print(f"💊{玩家角色.名稱}恢復了{補血量}點生命值!")

    elif 行動=="4":
        玩家角色.使用技能(敵人角色)

    elif 行動=="5":
        玩家角色.使用道具()

    else:
        print("❌無效的選項,請重新輸入!")
        continue
    if 敵人角色.是否存活():
        敵人角色.攻擊玩家(玩家角色)

    if 玩家角色.是否存活():
        print(f"\n🎉{玩家角色.名稱}成功擊敗了{敵人角色.名稱}")
        玩家角色.獲得經驗值(敵人角色.獎勵經驗)
    if random.random()<0.5:
        道具={"名稱":"小型生命藥水💊","效果":"恢復20點生命值","類型":"恢復","數值":20}
        玩家角色.背包.append(道具)
        print(f"🎁擊敗{敵人角色.名稱}後獲得道具:{道具['名稱']}!")
    else:
        print(f"\n💀{玩家角色.名稱}被{敵人角色.名稱}打敗了,遊戲結束!")
        break
    if 玩家角色.是否存活():
        print(f"\n🏆恭喜{玩家角色.名稱}成功通關!")
    else:
        print("\n💡請再接再厲!")
#啟動遊戲
if __name__=="__main__":
     主程式()
```
