```python
import random
import time

def show_holes(mole_position, remaining_time):
    """顯示地鼠洞和遊戲狀態"""
    print(f"\n🕒 剩餘時間: {int(remaining_time)}秒 | 🏆 當前分數: {score}")
    print("打地鼠！輸入 1~9 的數字來敲地鼠：")
    for i in range(1, 10):
        if i == mole_position:
            print("[🐭]", end=" ")
        else:
            print("[🕳️]", end=" ")
        if i % 3 == 0:
            print()  # 換行

def get_user_input(timeout, end_time):
    """獲取用戶輸入，有超時限制"""
    start = time.time()
    while True:
        # 檢查是否超時
        if time.time() > end_time:
            return None, True  # 遊戲時間結束

        # 檢查是否輸入超時
        if time.time() - start > timeout:
            return None, False  # 輸入超時但遊戲未結束

        # 嘗試獲取輸入
        try:
            if time.time() - start > timeout:
                return None, False

            user_input = input("你要打哪一個洞？(1~9): ")
            if user_input.lower() == "q":
                return None, True  # 玩家退出

            if not user_input.isdigit():
                print("⚠️ 請輸入數字！")
                continue

            choice = int(user_input)
            if 1 <= choice <= 9:
                return choice, False
            else:
                print("❌ 請輸入 1~9 的數字！")
        except KeyboardInterrupt:
            return None, True  # 玩家中斷遊戲

def main():
    global score
    score = 0
    total_time = 30  # 遊戲時間（秒）
    end_time = time.time() + total_time
    difficulty = 10  # 初始難度 (地鼠出現時間間隔)

    print("="*40)
    print("🎮 文字版打地鼠遊戲")
    print("="*40)
    print("遊戲規則:")
    print("- 你有 30 秒時間打地鼠")
    print("- 地鼠會隨機出現在 1~9 的洞中")
    print("- 輸入對應的洞號擊打地鼠")
    print("- 擊中得分，錯過扣分")
    print("- 按 'q' 可隨時退出遊戲")
    print("="*40)
    input("按 Enter 開始遊戲...")

    while True:
        remaining = end_time - time.time()
        if remaining <= 0:
            break

        # 隨機選擇地鼠位置
        mole_position = random.randint(1, 9)
        show_holes(mole_position, remaining)

        # 獲取玩家輸入
        choice, game_over = get_user_input(difficulty, end_time)

        if game_over:
            break

        if choice is None:  # 輸入超時
            print("⏰ 太慢了，地鼠跑掉了！-1分")
            score = max(0, score - 1)
        elif choice == mole_position:
            print("🎯 打中了！+1分")
            score += 1
            # 增加難度（減少下次出現時間）
            difficulty = max(0.5, difficulty * 0.95)
        else:
            print(f"❌ 沒打中！正確位置是 {mole_position} -1分")
            score = max(0, score - 1)
            # 降低難度
            difficulty = min(2.5, difficulty * 1.05)

        # 顯示當前分數和難度
        print(f"當前分數: {score} | 難度: {difficulty:.2f}秒")
        time.sleep(0.5)  # 稍微停頓

    print("\n" + "="*40)
    print(f"🎉 遊戲結束！最終分數: {score}")

    # 評分系統
    if score >= 15:
        print("🌟 太厲害了！你是打地鼠大師！")
    elif score >= 10:
        print("👍 表現不錯！再練習一下會更好！")
    elif score >= 5:
        print("😊 還可以，繼續加油！")
    else:
        print("🤔 需要多練習哦！")
    print("="*40)

if __name__ == "__main__":
    score = 0  # 全局分數變量
    main()
```
